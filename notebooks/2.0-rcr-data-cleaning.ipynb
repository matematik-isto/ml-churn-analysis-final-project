# Notebook 02 - Data Cleaning

## Objetivo

Generar datasets intermedios listos para ingeniería de características (feature engineering).

## Entrada
- `data/raw/contract.csv`
- `data/raw/personal.csv`
- `data/raw/internet.csv`

```python
## Inicialización
import pandas as pd
```

```python
## Cargar datos
raw_path = '../data/raw/' # Directorio para datos raw
df_contract = pd.read_csv(raw_path+'contract_raw.csv')
df_personal = pd.read_csv(raw_path+'personal_raw.csv')
df_internet = pd.read_csv(raw_path+'internet_raw.csv')
df_phone = pd.read_csv(raw_path+'phone_raw.csv')
```

### Archivo 1 (contract.csv)

```python
# Convertir tipo de dato a las columnas de fechas 
df_contract = pd.read_csv(raw_path+'contract.csv', parse_dates=['BeginDate','EndDate'])
```

```python
df_contract.isna().sum()
```

```python
# Validar cambio de tipo
df_contract.dtypes
```

```python
# Estandarizar nombres de columnas
df_contract.rename(columns={'customerID': 'customer_id', 'BeginDate':'begin_date', 'EndDate':'end_date', 
                            'Type':'type', 'PaperlessBilling':'paperless_billing', 'PaymentMethod':'payment_method',
                            'MonthlyCharges':'monthly_charges', 'TotalCharges':'total_charges'}, inplace=True)
```

```python
# Inspección de algunos elementos para tener mayor familaridad
df_contract.head()
```

```python
# Buscar duplicados totales
df_contract.duplicated().sum()
```

No hay duplicados. A continuación se investiga si hay valores incongruentes en alguna característica.

```python
# customer_id - al ser id no debería tener duplicados
df_contract['customer_id'].duplicated().sum()
```

```python
# begin_date
print("periodo del ",df_contract['begin_date'].min(), "al ",df_contract['begin_date'].max())
```

```python
# begin_date
df_contract['begin_date'].value_counts()
```

```python
# end_date
df_contract['end_date'].value_counts()
```

De los cinco distintos valores cuatro son fechas, al inicio de algunos meses, y existe el valor 'No'. La proporción de contratos terminados es, gruesamente, del 25%

```python
# type
df_contract['type'].value_counts()
```

```python
# paperless_billing
df_contract['paperless_billing'].value_counts()
```

```python
# payment_method
df_contract['payment_method'].value_counts()
```

```python
# monthly_charges
df_contract['monthly_charges'].describe()
```

```python
# total_charges
df_contract['total_charges'].sort_values().unique()
```

```python
# investigar cómo se ven los datos cuando total_charges = ' '
df_contract[df_contract['total_charges']==' '].head(10)
```

Se observa que todos los datos corresponden a contratos que inician al final del horizonte de tiempo, es decir contrantos nuevos y por tanto no han recibido cargos. Entonces se pueden reemplazar  (' ' $\rightarrow$ 0)

```python
# Validar observación
df_contract.loc[df_contract['total_charges']==' ','begin_date'].unique()
```

```python
# Se reemplazar cadenas en blanco por cero
df_contract['total_charges'] = df_contract['total_charges'].replace(' ', 0)
```

```python
# Cambiar el tipo de dato
df_contract['total_charges'] = df_contract['total_charges'].astype("float")
```

```python
df_contract.dtypes
```

### Segundo archivo (personal.csv)

```python
# Estandarizar nombres de columnas
df_personal.rename(columns={'customerID': 'customer_id', 'SeniorCitizen':'senior_citizen',
                            'Partner':'partner', 'Dependents':'dependents'}, inplace=True)
```

```python
# Buscar duplicados totales
df_personal.duplicated().sum()
```

```python
# Analizar valores de cada columna, ya sabemos que no hay valores ausentes None
# customer_id - al ser id no debería tener duplicados
df_personal['customer_id'].duplicated().sum()
```

```python
# gender
df_personal['gender'].value_counts()
```

```python
# senior_citizen

# Reemplazar números por Yes/No para seguir el mismo patrón que el resto de variables binarias
df_personal['senior_citizen'] = df_personal['senior_citizen'].apply(lambda x: 'Yes' if x ==1 else 'No')
df_personal['senior_citizen'].value_counts()
```

```python
# dependents
df_personal['dependents'].value_counts()
```
